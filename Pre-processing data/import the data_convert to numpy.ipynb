```python
import numpy as np
import os
import matplotlib.pyplot as plt

def load_seismic_dat(file_path, shape=None, dtype=np.float32):
    """Load a binary .dat file and reshape it into a 2D NumPy array if shape is provided."""
    data = np.fromfile(file_path, dtype=dtype)  # Read binary data
    
    # Automatically detect shape if not provided
    if shape is None:
        size = int(np.sqrt(data.shape[0]))  # Assume square shape if not specified
        shape = (size, size)
    
    try:
        data = data.reshape(shape)  # Reshape into 2D
    except ValueError:
        print(f"⚠️ Warning: {file_path} has {data.shape[0]} elements, incompatible with {shape}. Skipping...")
        return None  # Skip this file if shape mismatch
    
    return data

def convert_folder_to_npy(input_folder, output_folder, shape=None):
    """Convert all .dat files in a folder to .npy format."""
    os.makedirs(output_folder, exist_ok=True)  # Create output folder if it doesn't exist
    
    files = sorted([f for f in os.listdir(input_folder) if f.endswith('.dat')])
    
    if not files:
        print("❌ No .dat files found in the folder!")
        return

    print(f"🔄 Converting {len(files)} .dat files from {input_folder} to {output_folder}...")

    for file in files:
        file_path = os.path.join(input_folder, file)
        data = load_seismic_dat(file_path, shape)
        
        if data is not None:  # Only save if data is valid
            npy_filename = os.path.splitext(file)[0] + ".npy"  # Change extension
            npy_path = os.path.join(output_folder, npy_filename)
            np.save(npy_path, data)  # Save as NumPy format
            print(f"✅ Converted {file} → {npy_filename}")

# Example Usage
#convert_folder_to_npy("/home/g202393750/Documents/Superresolution/Test folder/", "/home/g202393750/Documents/Superresolution/Test folder/numpy", shape=(592, 400))
```
